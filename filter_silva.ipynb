```python
import numpy as np
from Bio import SeqIO
from collections import defaultdict
```

```python
fn = '../16S-mutations/SILVA_138.2_SSURef_NR99_tax_silva.fasta'
```

```python
# get list of seqs in each taxonomic group
level = 2
seqs = defaultdict(list)
for record in SeqIO.parse(fn, 'fasta'):
    # get taxonomic group
    tax = record.description.split(' ', 1)[1]
    tax = ';'.join(tax.split(';',level)[:level])
    # update
    seqs[tax].append(record)
```

```python
# get random seq in each group
sampled_seqs = []
for tax in seqs.keys():
    records = seqs[tax]
    i = np.random.randint(0,len(records))
    sampled_seqs.append(records[i])
```

```python
# output fasta
ofn = 'SILVA_SSU_tiny.fasta'
with open(ofn, 'w') as f:
    SeqIO.write(sampled_seqs, f, 'fasta')
```
